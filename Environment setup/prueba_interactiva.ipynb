# Prueba interactiva de ENV_RL

Selecciona **Python 3.10 (ENV_RL - CPU)** como kernel y ejecuta las celdas en orden con **Shift + Enter**. Este notebook comprueba el entorno y muestra cálculos, una gráfica y un entrenamiento breve de RL.


```python
import sys
from importlib import import_module
from importlib.metadata import version

print("Python:", sys.version.split()[0])
print("Ejecutable:", sys.executable)
if sys.version_info[:2] != (3, 10):
    raise RuntimeError("Selecciona el kernel ENV_RL con Python 3.10.")

esperadas = {
    "numpy": "1.26.4", "scipy": "1.14.1", "pandas": "2.2.3",
    "meshio": "5.3.5", "control": "0.10.1", "torch": "2.9.1",
    "gymnasium": "1.0.0", "stable-baselines3": "2.9.2",
}
for paquete, esperada in esperadas.items():
    modulo = "stable_baselines3" if paquete == "stable-baselines3" else paquete
    import_module(modulo)
    instalada = version(paquete)
    print(f"{paquete}: {instalada}")
    if instalada != esperada:
        raise RuntimeError(f"{paquete}: se esperaba {esperada}.")

import torch
if torch.version.cuda is not None:
    raise RuntimeError("Este setup requiere la compilacion de PyTorch para CPU.")
print("Compilacion de CPU confirmada.")
```

# NumPy, SciPy, pandas y Matplotlib

La integral de sin(x) entre 0 y pi debe ser cercana a **2**. La gráfica aparecerá debajo de la celda.


```python
%matplotlib inline
import numpy as np
import pandas as pd
from scipy.integrate import trapezoid
import matplotlib.pyplot as plt

x = np.linspace(0, np.pi, 200)
y = np.sin(x)
datos = pd.DataFrame({"x": x, "sin_x": y})

print("Integral aproximada:", trapezoid(y, x))
print(datos.head())

plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.grid(True)
plt.show()
```

# Stable-Baselines3 en CPU

Entrena PPO durante 128 pasos para comprobar que el conjunto funciona. Esta prueba breve no evalúa la calidad del agente. Si reinicias el kernel, vuelve a ejecutar las celdas anteriores.


```python
import gymnasium as gym
from stable_baselines3 import PPO

env = gym.make("CartPole-v1")
try:
    modelo = PPO(
        "MlpPolicy", env, device="cpu",
        n_steps=64, batch_size=32, n_epochs=1,
        seed=42, verbose=0,
    )
    modelo.learn(total_timesteps=128)
finally:
    env.close()

print("PRUEBA COMPLETADA: entrenamiento en CPU correcto.")
```
